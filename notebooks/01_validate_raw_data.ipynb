```python
import pandas as pd
import geopandas as gpd
from pathlib import Path

RAW = Path("../data/raw")

# Real raw-data filenames (glob-discovered, same as build scripts).
sales_path = next((RAW / "sales").glob("NYC_Citywide_Annualized_Calendar_Sales_Update_*.csv"))
pluto_path = next((RAW / "pluto").glob("Primary_Land_Use_Tax_Lot_Output_(PLUTO)_*.csv"))

sales = pd.read_csv(sales_path, nrows=5)
pluto = pd.read_csv(pluto_path, nrows=5)
council = gpd.read_file(RAW / "election_districts" / "NYC_City_Council_Districts.shp")
ed_path = next((RAW / "election_districts").glob("geo_export_*.shp"))
ed = gpd.read_file(ed_path)
elections = pd.read_csv(RAW / "election_results" / "ed_results_2021_mayor.csv", nrows=5)

print("Sales:", sales.shape, sales_path.name)
print("PLUTO:", pluto.shape, pluto_path.name)
print("Council:", council.shape)
print("ED:", ed.shape, ed_path.name)
print("Elections:", elections.shape)
```
