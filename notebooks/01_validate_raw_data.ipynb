```python
import pandas as pd
import geopandas as gpd
from pathlib import Path

RAW = Path("../data/raw")

sales = pd.read_csv(RAW / "sales" / "sales.csv", nrows=5)
pluto = pd.read_csv(RAW / "pluto" / "pluto.csv", nrows=5)
council = gpd.read_file(RAW / "council_districts")
ed = gpd.read_file(RAW / "election_districts")
elections = pd.read_csv(RAW / "election_results" / "elections.csv", nrows=5)

print("Sales:", sales.shape)
print("PLUTO:", pluto.shape)
print("Council:", council.shape)
print("ED:", ed.shape)
print("Elections:", elections.shape)
```
